```python
from argopy import ArgoIndex
idx = ArgoIndex(index_file="bgc-b").load()
df = idx.to_dataframe()
print(df.head())
print(df.columns)
```

```python
#all floats locations

import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.scatter(df["longitude"], df["latitude"], c="blue", s=5, alpha=0.5)
plt.title("Global Argo BGC Float Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

```python
#Indian Ocean floats locations
indian_df = df[(df["longitude"].between(40, 100)) & 
               (df["latitude"].between(-30, 30))]

print(indian_df.head())
```
