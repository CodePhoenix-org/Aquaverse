```python
import pandas as pd
import plotly.graph_objects as go

# Replace this with your actual data loading from 'argo_labeled.csv'
# df = pd.read_csv('argo_labeled.csv')
# For demonstration, a dummy DataFrame is used:
data = {
    'Temperature': [0.5, -0.2, 1.2, -0.8, 1.5, 0.1, -0.5, 0.9, 1.1, -0.3, 0.6, 0.7, 0.8, 0.9, 1.0, 1.1, 1.2, 1.3],
    'Salinity': [34.5, 35.0, 34.8, 35.5, 34.2, 35.1, 35.8, 34.9, 35.2, 35.6, 34.6, 34.7, 34.8, 34.9, 35.0, 35.1, 35.2, 35.3],
    'Oxygen': [0.8, -0.4, 0.6, -0.7, 0.9, -0.1, -0.6, 0.2, 0.7, -0.2, 0.5, 0.4, 0.3, 0.2, 0.1, 0.0, -0.1, -0.2],
    'Anomaly_Status': ['Normal', 'Anomaly', 'Normal', 'Anomaly', 'Normal', 'Anomaly', 'Anomaly', 'Normal', 'Normal', 'Anomaly', 'Normal', 'Normal', 'Normal', 'Normal', 'Normal', 'Normal', 'Normal', 'Normal']
}
df = pd.DataFrame(data)

# Create two separate DataFrames for 'Normal' and 'Anomaly' points
df_normal = df[df['Anomaly_Status'] == 'Normal']
df_anomaly = df[df['Anomaly_Status'] == 'Anomaly']

# Create the 3D scatter plot
fig = go.Figure()

# Add the 'Normal' data as blue circles
fig.add_trace(go.Scatter3d(
    x=df_normal['Temperature'],
    y=df_normal['Salinity'],
    z=df_normal['Oxygen'],
    mode='markers',
    marker=dict(
        size=5,
        color='blue',
        symbol='circle',
        opacity=0.8
    ),
    name='Normal'
))

# Add the 'Anomaly' data as red diamonds
fig.add_trace(go.Scatter3d(
    x=df_anomaly['Temperature'],
    y=df_anomaly['Salinity'],
    z=df_anomaly['Oxygen'],
    mode='markers',
    marker=dict(
        size=7,
        color='red',
        symbol='diamond',
        opacity=0.8
    ),
    name='Anomaly'
))

# Update the layout for a better view
fig.update_layout(
    title='3D: Temperature (x), Salinity (y), Oxygen (z), Anomaly (Color/Symbol)',
    scene=dict(
        xaxis_title='Temperature (°C)',
        yaxis_title='Salinity',
        zaxis_title='Oxygen',
    ),
    height=700
)

fig.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import plotly.express as px
import plotly.graph_objects as go
import numpy as np

# Replace this with your actual data loading from 'argo_labeled.csv'
# df = pd.read_csv('argo_labeled.csv')
# For demonstration, a dummy DataFrame is used:
np.random.seed(42)
num_points = 500
data = {
    'Longitude': np.random.uniform(-2.5, 2.5, num_points),
    'Latitude': np.random.uniform(-1, 1, num_points),
    'Depth': np.random.uniform(0, -2.5, num_points),  # Depth is negative
    'Temperature': np.random.uniform(-1, 1.5, num_points),
    'Anomaly_Status': ['Normal'] * int(num_points * 0.8) + ['Anomaly'] * int(num_points * 0.2)
}
df = pd.DataFrame(data)

# Create the 3D scatter plot
fig = px.scatter_3d(df,
                    x='Longitude',
                    y='Latitude',
                    z='Depth',
                    color='Temperature',
                    symbol='Anomaly_Status',
                    color_continuous_scale=px.colors.sequential.Viridis_r,
                    title='3D Ocean Data: Temperature (Color), Anomaly (Symbol), Temperature (Size)',
                    labels={'Temperature': 'Temperature Anomaly (°C)'})

# Update marker sizes and opacity
fig.update_traces(marker=dict(
    size=5,
    opacity=0.7
))

# Manually map Temperature values to colors from the color scale
norm = plt.Normalize(df['Temperature'].min(), df['Temperature'].max())
mapper = cm.ScalarMappable(norm=norm, cmap=cm.viridis_r)
colors = [f'rgb({int(c[0]*255)}, {int(c[1]*255)}, {int(c[2]*255)})' for c in mapper.to_rgba(df['Temperature'])]


# Add vertical lines (stems) from the points down to a baseline
for i in range(len(df)):
    fig.add_trace(go.Scatter3d(
        x=[df.iloc[i]['Longitude'], df.iloc[i]['Longitude']],
        y=[df.iloc[i]['Latitude'], df.iloc[i]['Latitude']],
        z=[df.iloc[i]['Depth'], df['Depth'].min() - 0.2],  # Draw line to a bit below the min depth
        mode='lines',
        line=dict(
            color=colors[i], # Use the manually mapped color
            width=2
        ),
        showlegend=False
    ))

# Update the layout
fig.update_layout(
    scene=dict(
        zaxis_title='Depth',
    ),
    height=700
)

fig.show()
```
